```python
# PROGRAM THAT CHECKS THE GIVEN NUMBER IS PALINDROME OR NOT USING FUNCTION
def palindrome(n):
  num=n
  rev=0
  while(n>0):
    digit=n%10
    rev=(rev*10)+digit
    n=n//10
  if rev==num:
      print("palindrome")
  else:
      print("not palindrome")
n=int(input())
palindrome(n)
m=int(input())
palindrome(m)
```

```python
# program on reverse of a string
def reverse(s):
  rev=""
  for i in range(len(s)-1,-1,-1):
    rev=rev+s[i]
  return rev
print(reverse("helllo"))
```

```python
# example on default arguments
def print_msg(name,branch="ai&ds"):
  print("the name is",name,"the branch is",branch)
print_msg("sai")
```

```python
# example on keyword arguments
def print_msg(name,branch):
  print("the name is",name,"the branch is",branch)
print_msg(branch="cse",name="sai")
```

```python
# example on variable length
def add(*num):
  print(num)
  sum=0
  for i in num:
    sum+=i
  print(sum)
add(1,2,3,4,5,6)
```
