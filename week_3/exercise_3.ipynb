```python
try:
  n = int(input("Enter a number for n: "))
except ValueError:
  print("Invalid input or no input provided. Defaulting n to 5.")
  n = 5 # Default value if input is invalid or empty

for i in range(1,11):
  print(f"n*i={n*i}")
```

```python
n=int(input())
for i in range(2,n):
  if(n%i==0):
    print("not prime")
    break
else:
  print("prime")
```

```python
# print prime numbers between two numbers
n1=int(input())
n2=int(input())
for i in range(n1,n2+1):
  for j in range(2,i):
    if(i%j==0):
      break
  else:
   print(i,end=" ")

```

```python

# print sum of digits of a given number
n=int(input())
sum=0
while(n!=0):
  rem=n%10
  sum=sum+rem
  n=n//10
print(sum)


```
